```python
import ipywidgets as widgets
from IPython.display import display
import multiprocessing
import time
import numpy as np

# Create a dictionary for hyperparameters
hyperparameters = {f"gpu_{i}": False for i in range(10)}

# Create checkboxes for each hyperparameter
checkboxes = [widgets.Checkbox(value=False, description=f'gpu_{i}', indent=False) for i in range(10)]

manager = multiprocessing.Manager()
available_devices = manager.list()
device_job_list = manager.list([[] for i in range(10)])

# Define a function for the subprocess
def subprocess_func(param_id, device_id):
    for _ in range(10):
        time.sleep(1)

# define a function to spawn new subprocesses
def spawn_subprocess(available_devices, device_job_list):
    job_list = [i for i in range(100)]
    # create a thread which uses checkboxes to determine the available devices

    while job_list is not None:
        job_id = job_list.pop()
        device_id = np.choice(available_devices)
        p = multiprocessing.Process(target=subprocess_func, args=(job_id, device_id))
        p.start()
    return p

forker = multiprocessing.Process(target=spawn_subprocess, args=(available_devices, device_job_list))
forker.start()

# Create events for each subprocess
# events = [multiprocessing.Event() for _ in range(10)]

# # Start the subprocesses
# p = []
# for i in range(10):
#     p.append(multiprocessing.Process(target=subprocess_func, args=(i, events[i], displays[i])))
#     p[i].start()

# Define a function to handle checkbox changes
def checkbox_change(change):
    if change["name"] == "value" and (change["new"] != change["old"]):
        device_id = int(change["owner"].description.split("_")[1])
        if change["new"]:
            available_devices.append(device_id)
        else:
            available_devices.remove(device_id)

# Add the checkbox change handler to each checkbox
for checkbox in checkboxes:
    checkbox.observe(checkbox_change)

# Display the checkboxes
display(*checkboxes)

did = display(display_id=True)

while forker.is_alive():
    did.display("Available devices: " + str(available_devices))
# for i in range(10):
#     p[i].join()
```
