```python
import os
from xgraph.definitions import ROOT_DIR
from pathlib import Path
# from openpyxl import Workbook, load_workbook
from IPython.display import display
import pandas as pd
import sys
import time
import sqlite3
import numpy as np


```

```python
df = pd.read_excel(Path(ROOT_DIR, 'results_backup', 'GCN_GIN_PL.xlsx'), sheet_name='GCN', index_col=[0, 1], header=[0, 1])
```

```python
# df = df['GCN'].unstack(0)
df = df.unstack().unstack(0).unstack(0)
```

```python
# df.index = df.index.append(pd.Index([(np.array(sp * 0.1).round(1), 'abc') for sp in range(5, 10)]))
# Add one index into the multi index
df = pd.concat([df, pd.DataFrame(index=pd.MultiIndex.from_tuples([(np.array(sp * 0.1).round(1), 'abc') for sp in range(5, 10)]))]).sort_index()
```

```python
# df.sort_index(inplace=True)
pd.concat([df, pd.DataFrame(index=pd.MultiIndex.from_product([[0.5, 0.6, 0.7, 0.8, 0.9], ['new_method']]), columns=pd.MultiIndex.from_tuples([('ff', metric) for metric in ['Fidelity+', 'Fidelity-']]))]).sort_index().sort_index(axis=1)
```

```python
pd.DataFrame(index=pd.MultiIndex.from_product([[0.5, 0.6, 0.7, 0.8, 0.9], ['new_method']]), columns=pd.MultiIndex.from_tuples([('ff', metric) for metric in ['Fidelity+', 'Fidelity-']]))
```

```python
with sqlite3.connect(Path(ROOT_DIR, 'quantitative_results', 'GCN_GIN_PL.db')) as conn:
    df['GCN'].to_sql('GCN_PL', conn, if_exists='replace')
    df['GIN'].to_sql('GIN_PL', conn, if_exists='replace')
```

```python
with sqlite3.connect(Path(ROOT_DIR, 'quantitative_results', 'GCN_GIN_PL.db')) as conn:
    df_sql = pd.read_sql('select * from GCN_PL', conn, index_col=['model', 'sparsity'])
    display(df_sql)
```

```python
pd.concat([df, pd.DataFrame(index=pd.MultiIndex.from_product([df.index.levels[0], ['new_method']], names=df.index.names), columns=pd.MultiIndex.from_tuples([('ff', metric) for metric in ['Fidelity+', 'Fidelity-']], names=df.columns.names))]).sort_index().sort_index(axis=1)
```
